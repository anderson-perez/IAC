# Introdução ao Aprendizado Não Supervisionado

Ao contrário do aprendizado supervisionado, o **aprendizado não supervisionado** tem apenas as características:

$$
X =
\begin{bmatrix}
x_{11} & x_{12} & \cdots & x_{1m}\\
x_{21} & x_{22} & \cdots & x_{2m}\\
\vdots & \vdots & \ddots & \vdots\\
x_{n1} & x_{n2} & \cdots & x_{nm}
\end{bmatrix}
$$

Não existe uma coluna $(y)$ dizendo ao modelo qual é a resposta correta.

O objetivo passa a ser **explorar a estrutura dos próprios dados**, procurando regularidades que possam ajudar a compreendê-los.

O aprendizado não supervisionado trabalha com dados para os quais **não se conhece previamente a resposta desejada**. Em vez de aprender diretamente uma relação entre entradas e rótulos, o modelo tenta descobrir estruturas existentes nos próprios dados.

Três características importantes do aprendizado não supervisionado:

1. **semelhança e agrupamento** — identificar observações que apresentam características semelhantes;
2. **representação** — descrever dados complexos utilizando formas mais compactas;
3. **anomalias** — identificar observações que se diferenciam significativamente do comportamento predominante.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)
```

## Exemplo Numérico

Considerando um conjunto de clientes de uma loja utilizando somente as características:

- número de compras realizadas por mês;
- valor médio gasto em cada compra.

Não existe nenhuma informação sobre qual categoria cada cliente se enquadra.

```python
clientes = pd.DataFrame({
    "compras_mes": [2, 3, 4, 3, 5, 11, 12, 10, 13, 11, 5, 6, 7, 6, 8],
    "gasto_medio": [42, 38, 50, 45, 47, 145, 155, 138, 162, 150, 88, 95, 92, 102, 98]
})

clientes
```

```python
plt.figure(figsize=(7, 5))
plt.scatter(clientes["compras_mes"], clientes["gasto_medio"], s=70)
plt.xlabel("Compras por mês")
plt.ylabel("Gasto médio")
plt.title("Dados sem rótulos")
plt.grid(True, alpha=0.3)
plt.show()
```

O gráfico sugere que existem **regiões com observações semelhantes**.

Essa é a ideia central do aprendizado não supervisionado:

> Ao invés de perguntar “qual é a resposta correta?”, em aprendizado não supervisionado é perguntado “que estrutura existe nesses dados?”.

É importante observar que os grupos percebidos visualmente não são rótulos verdadeiros. Eles são apenas padrões que parecem existir na distribuição dos dados.

## Semelhança entre Observações

Para identificar estrutura, é necessário alguma maneira de dizer se duas observações são **parecidas**.

Uma possibilidade simples é representar cada observação como um vetor:

$$
\mathbf{x}_i = [x_{i1},x_{i2},\ldots,x_{im}]
$$

e medir a distância entre dois vetores.

Uma medida comum é a **distância Euclidiana**:

$$
d(\mathbf{x},\mathbf{z}) =
\sqrt{\sum_{j=1}^{m}(x_j-z_j)^2}
$$

Quanto menor a distância, mais próximas são as observações nesse espaço de características.

```python
def distancia_euclidiana(a, b):
    a = np.asarray(a)
    b = np.asarray(b)
    return np.sqrt(np.sum((a - b)**2))

cliente_1 = clientes.iloc[0].values
cliente_2 = clientes.iloc[1].values
cliente_8 = clientes.iloc[7].values

print("Distância entre clientes 1 e 2:", distancia_euclidiana(cliente_1, cliente_2))
print("Distância entre clientes 1 e 8:", distancia_euclidiana(cliente_1, cliente_8))
```

## A Importância da Escala das Características

Considerando as características:

- idade: valores aproximadamente entre 18 e 70;
- renda anual: valores aproximadamente entre 20.000 e 200.000.

Se for calculada a distância, a característica com valores numericamente maiores pode dominar o resultado.

Por isso, em muitos problemas é importante colocar as características em **escalas comparáveis**.

Uma transformação de escala é a padronização:

$$
z = \frac{x-\mu}{\sigma}
$$

onde $(\mu)$ é a média e $(\sigma)$ é o desvio-padrão da característica.

```python
X = clientes.values.astype(float)

media = X.mean(axis=0)
desvio = X.std(axis=0)

X_padronizado = (X - media) / desvio

pd.DataFrame(
    X_padronizado,
    columns=["compras_mes_padronizado", "gasto_medio_padronizado"]
).head()
```

## Tarefas no Aprendizado Não-Supervisionado

* Agrupamento (*clustering*)
* Redução de dimensionalidade
* Detecção de anomalias
* Regras de associação
* Segmentação

## Clusterização

O **clustering** procura dividir os dados em grupos de forma que:

- elementos do mesmo grupo sejam semelhantes;
- elementos de grupos diferentes sejam distintos.

No clustering o objetivo é particionar um conjunto de dados em \(K\) grupos:

$
C_1, C_2, \dots, C_K
$

onde cada grupo $(C_k)$ contém pontos semelhantes entre si.



## Aplicação da Clusterização

* clientes com comportamento parecido;
* sensores com padrões similares;
* imagens com características semelhantes;
* medições de vibração de máquinas em condições similares.

```python
plt.figure(figsize=(7, 5))
plt.scatter(X_padronizado[:, 0], X_padronizado[:, 1], s=70)
plt.xlabel("Compras por mês (padronizado)")
plt.ylabel("Gasto médio (padronizado)")
plt.title("Estrutura dos dados em um espaço padronizado")
plt.axhline(0, linewidth=0.8)
plt.axvline(0, linewidth=0.8)
plt.grid(True, alpha=0.3)
plt.show()
```

### Agrupamento não é Classificação

Na **classificação**, as classes já existem nos dados de treinamento:

$$
(X,y)
$$

No **agrupamento**, as classes não são conhecidas:

$$
X
$$

Grupos podem ser encontrados a partir da estrutura dos dados. Uma pessoa pode analisar esses grupos e tentar interpretá-los.

Um grupo encontrado nos dados também não precisa corresponder necessariamente a uma categoria real do mundo.

## Redução de Dimensionalidade

Um conjunto de dados pode possuir dezenas, centenas ou milhares de características.

Uma observação por se representada como:

$$
\mathbf{x}=[x_1,x_2,\ldots,x_{100}]
$$

Visualizar diretamente um espaço com 100 dimensões é impossível.

Uma segunda família de técnicas não supervisionadas procura criar uma **representação com menos dimensões**, preservando informações relevantes da estrutura original.

Por exemplo:

$$
100\text{ características}
\longrightarrow
2\text{ ou }3\text{ novas características}
$$

A redução de dimensionalidade pode ser útil para:

- visualização;
- redução de redundância;
- compressão;
- pré-processamento;
- exploração dos dados.

## Detecção de Observações Incomuns

Outra tarefa possível é procurar exemplos que se comportam de maneira muito diferente da maioria.

Esses exemplos podem ser chamados de **anomalias** ou *outliers*.

Uma anomalia não significa necessariamente um erro. Ela pode representar:

- uma medição incorreta;
- uma falha de sensor;
- uma operação incomum;
- um comportamento raro;
- um evento relevante que merece investigação.

```python
x = np.array([1.0, 1.2, 0.9, 1.1, 1.3, 1.05, 0.95, 4.5])
y = np.array([1.1, 0.9, 1.0, 1.2, 1.05, 0.95, 1.15, 4.2])

plt.figure(figsize=(6, 5))
plt.scatter(x, y, s=80)

for i, (xi, yi) in enumerate(zip(x, y)):
    plt.annotate(str(i + 1), (xi, yi), xytext=(5, 5), textcoords="offset points")

plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.title("Uma observação distante das demais")
plt.grid(True, alpha=0.3)
plt.show()
```

No gráfico, a observação 8 está distante das demais. Isso é um **indício visual**, não uma conclusão automática de que o ponto seja inválido.

Em aplicações reais, a interpretação depende do contexto do problema.

## Avaliação de Modelos de Aprendizado Não Supervisionado

No aprendizado supervisionado, uma prevsão pode ser comparada com o valor verdadeiro:

$$
\hat{y} \quad \text{versus} \quad y
$$

No aprendizado não supervisionado, não existe um $(y)$ conhecido.

Por isso, a avaliação pode ser menos direta e depender de diferentes critérios, como:

- quão bem a estrutura encontrada representa os dados;
- separação ou coesão de possíveis grupos;
- quantidade de informação preservada em uma nova representação;
- estabilidade dos padrões encontrados;
- utilidade dos resultados para o problema real;
- conhecimento de especialistas do domínio.

Assim, um padrão matematicamente interessante nem sempre é necessariamente útil ou significativo para a aplicação.

## Aprendizado Supervisionado × Não Supervisionado

| Aspecto | Supervisionado | Não supervisionado |
|---|---|---|
| Dados de entrada | Características \(X\) | Características \(X\) |
| Rótulos \(y\) | Sim | Não |
| Objetivo | Aprender a prever uma saída | Descobrir estrutura nos dados |
| Exemplos de tarefas | Classificação e regressão | Agrupamento, representação e anomalias |
| Avaliação | Comparação com respostas conhecidas | Depende da estrutura e da aplicação |

As duas abordagens não são concorrentes. Em muitos projetos, técnicas não supervisionadas podem ser utilizadas para **explorar e compreender os dados antes de construir modelos supervisionados**.

## Aplicações das Técnicas de Aprendizado Não Supervisionado

O aprendizado não supervisionado pode ser utilizado em problemas como:

- segmentação de clientes;
- exploração de dados científicos;
- identificação de padrões de comportamento;
- análise de sensores;
- detecção de situações incomuns;
- organização de documentos;
- visualização de dados de alta dimensionalidade;
- descoberta de representações úteis para outras tarefas de IA.
