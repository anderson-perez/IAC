# Funções

Uma função associa uma entrada a uma saída:

$$
y=f(x)
$$

Em IA, um modelo pode ser visto de maneira geral como uma função:

$$
\hat y=f(\mathbf{x};\theta)
$$

onde:


$\mathbf{x}$: entrada;

$\theta$: parâmetros do modelo;

$\hat y$: saída estimada.


Por exemplo:

$$
f(x)=2x+1
$$

Pode ser considerada um **modelo extremamente simples** com parâmetros $(2)$ e $(1)$.

````python
def f(x):
    return 2*x + 1

x = np.linspace(-5, 5, 100)
y = f(x)

plt.figure(figsize=(7,4))
plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Exemplo de função")
plt.grid(True, alpha=0.3)
plt.show()
````

## Algumas Funções Importantes para a Inteligência Artificial

### Exponencial

$$
f(x)=e^x
$$

### Logaritmo natural

$$
f(x)=\ln(x)
$$

### Sigmoide

$$
\sigma(x)=\frac{1}{1+e^{-x}}
$$

A sigmoide transforma qualquer número real em um valor entre 0 e 1.

### ReLU

$$
\operatorname{ReLU}(x)=\max(0,x)
$$

Essas funções aparecem em modelos probabilísticos e redes neurais.

````python
x = np.linspace(-6, 6, 300)

sigmoid = 1/(1+np.exp(-x))
relu = np.maximum(0, x)

plt.figure(figsize=(7,4))
plt.plot(x, sigmoid, label="Sigmoide")
plt.plot(x, relu, label="ReLU")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

# Somatórios e Produtórios

A notação de somatório:

$
\sum_{i=1}^{n}x_i
$

significa:

$
x_1+x_2+\cdots+x_n
$

Ela aparece constantemente em funções de erro. Por exemplo, o erro quadrático médio:

$
MSE=
\frac{1}{n}
\sum_{i=1}^{n}
(y_i-\hat y_i)^2
$

O produtório:

$
\prod_{i=1}^{n}x_i
$

representa:

$
x_1x_2\cdots x_n
$

e aparece frequentemente em probabilidades conjuntas e funções de verossimilhança.

````python
valores = np.array([1, 2, 3, 4])

print("Somatório:", np.sum(valores))
print("Produtório:", np.prod(valores))

y = np.array([2, 4, 6])
y_pred = np.array([2.2, 3.5, 6.3])

mse = np.mean((y-y_pred)**2)
print("MSE:", mse)
````

## Operações com Vetores

Dados:

$$
\mathbf{a}=
\begin{bmatrix}a_1\\a_2\end{bmatrix},
\qquad
\mathbf{b}=
\begin{bmatrix}b_1\\b_2\end{bmatrix}
$$

é possível calcular a soma, subtração e multiplicação por escalar.

O **produto escalar** é especialmente importante:

$$
\mathbf{a}\cdot\mathbf{b}
=
\sum_i a_ib_i
$$

**Modelos lineares e neurônios artificiais**, por exemplo, utilizam diretamente essa operação:

$$
z=\mathbf{w}^\top\mathbf{x}+b
$$

````python
a = np.array([2., 3.])
b = np.array([4., 1.])

print("Soma dos vetores (a+b) =", a+b)
print("Soma dos vetores (a-b) =", a-b)
print("Multiplicação por escalar (2a) =", 2*a)
print("Produto escalar (a·b) =", np.dot(a,b))

x = np.array([0.5, 1.2, -0.7])
w = np.array([0.8, -0.4, 1.1])
bias = 0.2

z = np.dot(w, x) + bias
print("\nSaída linear wᵀx+b =", z)
````

# Normas e Distâncias

A norma Euclidiana de um vetor é:

$$
\|\mathbf{x}\|_2=
\sqrt{\sum_i x_i^2}
$$

A distância Euclidiana entre dois pontos é obtida da seguinte forma:

$$
d(\mathbf{x},\mathbf{y})
=
\|\mathbf{x}-\mathbf{y}\|_2
$$

Outras normas importantes:

$$
\|\mathbf{x}\|_1=\sum_i|x_i|
$$

$$
\|\mathbf{x}\|_\infty=\max_i|x_i|
$$

Esses conceitos são importantes em **algoritmos baseados em distância, regularização e otimização**.

````python
x = np.array([3., 4.])
y = np.array([1., 1.])

print("Norma L2 de x:", np.linalg.norm(x))
print("Norma L1 de x:", np.linalg.norm(x, ord=1))
print("Distância entre x e y:", np.linalg.norm(x-y))
````

## Similaridade do Cosseno

Para comparar a **direção** de dois vetores, utiliza-se:

$$
\cos(\theta)
=
\frac{\mathbf{a}\cdot\mathbf{b}}
{\|\mathbf{a}\|\|\mathbf{b}\|}
$$

Valores próximos de 1 indicam direções semelhantes.

Essa medida é importante em **recuperação de informação, embeddings e processamento de linguagem natural**.

````python
a = np.array([1., 2., 3.])
b = np.array([2., 4., 5.])

cos = np.dot(a,b)/(np.linalg.norm(a)*np.linalg.norm(b))
print("Similaridade do cosseno:", cos)
````

# Matrizes

Uma matriz pode representar várias amostras simultaneamente:

$$
X=
\begin{bmatrix}
x_{11}&x_{12}&\cdots&x_{1m}\\
x_{21}&x_{22}&\cdots&x_{2m}\\
\vdots&\vdots&\ddots&\vdots\\
x_{n1}&x_{n2}&\cdots&x_{nm}
\end{bmatrix}
$$

Uma convenção comum em aprendizado de máquina é:

```text
linhas  → amostras
colunas → características
```

Se \(X\) possui dimensão $n\times m$, então, exitems $(n)$ amostras e $(m)$ características.

````python
X = np.array([
    [1.0, 2.0, 0.5],
    [1.5, 1.8, 0.7],
    [3.0, 3.5, 1.2],
    [2.5, 3.0, 1.0]
])

print(X)
print("\nFormato:", X.shape)
print("Amostras:", X.shape[0])
print("Características:", X.shape[1])
````

## Multiplicação de Matrizes

Se:

$$
A\in\mathbb{R}^{m\times n}
\quad\text{e}\quad
B\in\mathbb{R}^{n\times p},
$$

então:

$$
AB\in\mathbb{R}^{m\times p}.
$$

A multiplicação matricial é uma das operações fundamentais das **redes neurais artificiais**.

Uma camada pode ser escrita como:

$$
\mathbf{z}=W\mathbf{x}+\mathbf{b}
$$

````python
W = np.array([
    [0.2, 0.8, -0.5],
    [1.0, -0.3, 0.4]
])

x = np.array([1.0, 0.5, 2.0])
b = np.array([0.1, -0.2])

# Produto matricial (operador @)
z = W @ x + b

print("W shape:", W.shape)
print("x shape:", x.shape)
print("Saída:", z)
````

# Sistemas Lineares, Inversa e Pseudoinversa

Dado um sistema linear escrito como:

$$
A\mathbf{x}=\mathbf{b}
$$

Se \(A\) é quadrada e invertível:

$$
\mathbf{x}=A^{-1}\mathbf{b}
$$

Entretanto, em computação numérica normalmente evita-se calcular explicitamente a inversa quando for possível resolver o sistema diretamente.

Em problemas de aprendizado de máquina, também aparecem sistemas sem solução exata ou matrizes não quadradas. Nesses casos, a **pseudoinversa de Moore-Penrose** é útil:

$$
\mathbf{x}=A^+\mathbf{b}
$$

````python
A = np.array([[2., 1.],
              [1., 3.]])
b = np.array([5., 6.])

x = np.linalg.solve(A, b)

print("Solução:", x)
print("Verificação A dot x:", A @ x)
````

# Autovalores e Autovetores

Um vetor não nulo $\mathbf{v}$ é autovetor de $(A)$ quando:

$$
A\mathbf{v}=\lambda\mathbf{v}
$$

onde $(\lambda)$ é o autovalor correspondente.

Isso significa que a transformação representada por $(A)$ altera a escala do autovetor, mas não sua direção.

Autovalores e autovetores aparecem em:

- PCA (*Principal Componentes Analysis*);
- análise de estabilidade;
- métodos espectrais;
- cadeias de Markov;
- processamento de sinais.

````python
A = np.array([[3., 1.],
              [1., 3.]])

autovalores, autovetores = np.linalg.eig(A)

print("Autovalores:")
print(autovalores)

print("\nAutovetores:")
print(autovetores)
````

# Decomposição em Valores Singulares — SVD

Uma matriz pode ser decomposta como:

$$
A=U\Sigma V^\top
$$

Essa é a **Singular Value Decomposition (SVD)**.

Ela é fundamental em:

- redução de dimensionalidade;
- compressão;
- sistemas de recomendação;
- processamento de imagens;
- cálculo da pseudoinversa;
- análise de dados.

````python
A = np.array([
    [3., 2., 2.],
    [2., 3., -2.]
])

U, S, Vt = np.linalg.svd(A, full_matrices=False)

print("U =\n", U)
print("\nValores singulares =", S)
print("\nVᵀ =\n", Vt)

Sigma = np.diag(S)
print("\nReconstrução:\n", U @ Sigma @ Vt)
````

# Derivadas

A derivada mede a taxa de variação:

$$
f'(x)
=
\lim_{h\to0}
\frac{f(x+h)-f(x)}{h}
$$

Geometricamente, corresponde à inclinação da reta tangente.

Em IA, derivadas são fundamentais porque o treinamento de muitos modelos consiste em **minimizar uma função de perda**.

````python
f = lambda x: x**2

x = np.linspace(-4, 4, 200)
y = f(x)

x0 = 2
derivada = 2*x0
tangente = f(x0) + derivada*(x-x0)

plt.figure(figsize=(7,4))
plt.plot(x, y, label="f(x)=x²")
plt.plot(x, tangente, label="Tangente em x=2")
plt.scatter([x0], [f(x0)])
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

## Derivadas Parciais e Gradiente

Quando uma função depende de várias variáveis, como:

$$
f(x,y)=x^2+y^2
$$

calcula-se:

$$
\frac{\partial f}{\partial x}=2x
$$

e:

$$
\frac{\partial f}{\partial y}=2y
$$

O vetor contendo todas as derivadas parciais é o **gradiente**:

$$
\nabla f=
\begin{bmatrix}
\frac{\partial f}{\partial x}\\
\frac{\partial f}{\partial y}
\end{bmatrix}
$$

O gradiente aponta para a direção de maior crescimento local da função.

````python
def gradiente(x, y):
    return np.array([2*x, 2*y])

p = np.array([2., 1.])

print("Ponto:", p)
print("Gradiente:", gradiente(*p))
````

## Regra da Cadeia

O cálculo do algoritmo **backpropagation** é realizado com base na regra da cadeia.

Se:

$$
y=f(u)
$$

e:

$$
u=g(x),
$$

então:

$$
\frac{dy}{dx}
=
\frac{dy}{du}
\frac{du}{dx}
$$

Redes neurais são composições de muitas funções:

$$
\mathbf{x}
\rightarrow z_1
\rightarrow a_1
\rightarrow z_2
\rightarrow a_2
\rightarrow \hat y
$$

O backpropagation aplica repetidamente a regra da cadeia para calcular como cada parâmetro influencia o erro.

````python
# y = (3x + 1)^2
# u = 3x + 1
# y = u^2

x = 2.0
u = 3*x + 1

dy_du = 2*u
du_dx = 3

dy_dx = dy_du * du_dx

print("u =", u)
print("dy/du =", dy_du)
print("du/dx =", du_dx)
print("dy/dx =", dy_dx)
````

# Otimização e Descida do Gradiente

Considerando a necessidade de minimizar:

$$
J(w)=(w-3)^2
$$

A derivada é:

$$
J'(w)=2(w-3)
$$

A descida do gradiente atualiza:

$$
w_{k+1}
=
w_k-\eta J'(w_k)
$$

onde $(\eta)$ é a **taxa de aprendizado** (*learning rate*).

A ideia é mover os parâmetros na direção oposta ao gradiente.

````python
def J(w):
    return (w-3)**2

def dJ(w):
    return 2*(w-3)

w = -4.0
eta = 0.1

historico = []

for epoca in range(30):
    historico.append((epoca, w, J(w)))
    w = w - eta*dJ(w)

hist = pd.DataFrame(historico, columns=["Iteração", "w", "J(w)"])
display(hist.head(10))

plt.figure(figsize=(7,4))
plt.plot(hist["Iteração"], hist["J(w)"], marker="o")
plt.xlabel("Iteração")
plt.ylabel("J(w)")
plt.title("Descida do gradiente")
plt.grid(True, alpha=0.3)
plt.show()

print("w final:", w)
````

## A Taxa de Aprendizado

A escolha de $(\eta)$ é importante:

```text
η muito pequeno
    └── convergência lenta

η adequado
    └── convergência eficiente

η muito grande
    └── oscilações ou divergência
```

````python
def executar_gd(eta, passos=20):
    w = -4.0
    valores = []

    for _ in range(passos):
        valores.append(J(w))
        w -= eta*dJ(w)

    return valores

plt.figure(figsize=(7,4))

for eta in [0.02, 0.1, 0.4, 1.05]:
    plt.plot(executar_gd(eta), label=f"η={eta}")

plt.yscale("log")
plt.xlabel("Iteração")
plt.ylabel("J(w)")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

### Discreta

Assume valores contáveis.

Exemplo: número obtido em um dado.

### Contínua

Pode assumir valores em um intervalo contínuo.

# Distribuições de Probabilidade

Uma distribuição de probabilidades descreve como as probabilidades estão associadas aos possíveis valores de uma variável aleatória.

## Bernoulli

Representa experimentos binários:

$$
X\in\{0,1\}
$$

com:

$$
P(X=1)=p
$$

É natural para problemas binários.

## Normal

$$
X\sim\mathcal{N}(\mu,\sigma^2)
$$

Sua densidade é:

$$
f(x)=
\frac{1}{\sigma\sqrt{2\pi}}
e^{-\frac{(x-\mu)^2}{2\sigma^2}}
$$

A distribuição normal aparece em erros de medição, ruído, modelagem estatística e diversas aproximações.

````python
rng = np.random.default_rng(42)

amostras = rng.normal(loc=10, scale=1, size=10000)

plt.figure(figsize=(7,4))
plt.hist(amostras, bins=50, density=True)
plt.xlabel("x")
plt.ylabel("Densidade")
plt.title("Amostras de N(10, 1)")
plt.grid(True, alpha=0.3)
plt.show()

print("Média amostral:", np.mean(amostras))
print("Variância amostral:", np.var(amostras))
````

# Esperança e Variância

A esperança representa o valor médio teórico.

Para uma variável discreta:

$$
E[X]
=
\sum_x xP(X=x)
$$

A variância mede a dispersão:

$$
Var(X)
=
E[(X-E[X])^2]
$$

e o desvio-padrão:

$$
\sigma=\sqrt{Var(X)}
$$

Essas quantidades aparecem em estatística, inicialização de redes, normalização, métodos probabilísticos e aprendizado por reforço.

````python
x = np.array([1,2,3,4,5,6])
p = np.ones(6)/6

esperanca = np.sum(x*p)
variancia = np.sum((x-esperanca)**2*p)

print("E[X] =", esperanca)
print("Var(X) =", variancia)
print("Desvio-padrão =", np.sqrt(variancia))
````

# Probabilidade Condicional

A probabilidade condicional é:

$$
P(A\mid B)
=
\frac{P(A\cap B)}{P(B)}
$$

Ela responde:

> Qual é a probabilidade de $(A)$, sabendo que $(B)$ ocorreu?

Em IA, muitas tarefas podem ser interpretadas dessa maneira:

$$
P(\text{classe}\mid\text{características})
$$

Por exemplo:

$$
P(\text{falha}\mid
\text{temperatura, vibração, corrente})
$$

# Teorema de Bayes

O Teorema de Bayes é:

$$
P(H\mid E)
=
\frac{P(E\mid H)P(H)}
{P(E)}
$$

Cuja interpretação é:

- $(P(H))$: **prior**;
- $(P(E\mid H))$: **verossimilhança**;
- $(P(H\mid E))$: **posterior**.

Conceitualmente:

$$
\text{posterior}
\propto
\text{verossimilhança}
\times
\text{prior}
$$

O teorema de Bayes permite atualizar a crença sobre uma hipótese quando novas evidências são recebidas.

````python
# Exemplo: probabilidade de congestionamento dado que está chovendo

p_congestionamento = 0.30
p_normal = 0.70

p_chuva_dado_congestionamento = 0.80
p_chuva_dado_normal = 0.20

p_chuva = (
    p_chuva_dado_congestionamento*p_congestionamento
    + p_chuva_dado_normal*p_normal
)

posterior = (
    p_chuva_dado_congestionamento*p_congestionamento
    / p_chuva
)

print(f"P(Congestionamento | Chuva) = {posterior:.3f}")
````

# Estatística Descritiva

A estatística descritiva permite entender os dados antes de implementar e treinar um modelo de IA.

Medidas importantes:

### Média

$$
\bar{x}
=
\frac{1}{n}\sum_i x_i
$$

### Mediana

Valor central dos dados ordenados.

### Variância amostral

$$
s^2=
\frac{1}{n-1}
\sum_i(x_i-\bar{x})^2
$$

### Desvio-padrão

$$
s=\sqrt{s^2}
$$

Essas medidas ajudam a identificar localização, dispersão e possíveis valores atípicos.

````python
dados = np.array([10, 11, 9, 12, 10, 10, 11, 30])

print("Média:", np.mean(dados))
print("Mediana:", np.median(dados))
print("Variância amostral:", np.var(dados, ddof=1))
print("Desvio-padrão:", np.std(dados, ddof=1))
````

# Covariância e Correlação

A covariância mede como duas variáveis variam conjuntamente:

$$
Cov(X,Y)
=
E[(X-E[X])(Y-E[Y])]
$$

A correlação normaliza essa relação:

$$
\rho_{XY}
=
\frac{Cov(X,Y)}
{\sigma_X\sigma_Y}
$$

A correlação varia entre -1 e 1.

> Correlação não implica causalidade.

Essas medidas são importantes para análise exploratória, seleção de características e redução de dimensionalidade.

````python
rng = np.random.default_rng(3)

x = np.linspace(0, 10, 100)
y = 2*x + rng.normal(0, 2, 100)

print("Covariância:")
print(np.cov(x,y))

print("\nCorrelação:")
print(np.corrcoef(x,y))

plt.figure(figsize=(6,4))
plt.scatter(x,y)
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True, alpha=0.3)
plt.show()
````

# Padronização

Dados podem possuir escalas muito diferentes.

Exemplo:

```text
temperatura → 0 a 100
distância   → 0 a 10
RPM         → 0 a 8000
```

Uma transformação comum é o **z-score**:

$$
z=
\frac{x-\mu}{\sigma}
$$

Após a padronização, a variável possui aproximadamente média 0 e desvio-padrão 1.

Isso é particularmente importante em algoritmos sensíveis à escala e em otimização.

````python
x = np.array([10., 12., 15., 18., 20.])

mu = np.mean(x)
sigma = np.std(x)

z = (x-mu)/sigma

print("Original:", x)
print("Padronizado:", z)
print("Média:", np.mean(z))
print("Desvio-padrão:", np.std(z))
````

# Máxima Verossimilhança

A **verossimilhança** mede quão compatíveis determinados parâmetros são com os dados observados.

Se as observações são independentes:

$$
L(\theta)
=
\prod_{i=1}^{n}
P(x_i\mid\theta)
$$

Em vez de maximizar o produto, geralmente maximiza-se o logaritmo:

$$
\log L(\theta)
=
\sum_{i=1}^{n}
\log P(x_i\mid\theta)
$$

O logaritmo transforma produtos em somas e melhora a conveniência matemática e numérica.

Máxima verossimilhança está por trás de muitos modelos estatísticos e funções de perda utilizadas em aprendizado de máquina.

# Entropia

Na teoria da informação, a entropia mede a incerteza de uma distribuição:

$$
H(X)
=
-\sum_x p(x)\log_2p(x)
$$

Para um evento binário:

- $(p\approx0)$ ou $(p\approx1)$: pouca incerteza;
- $(p=0,5)$: máxima incerteza.

````python
def entropia_binaria(p):
    if p == 0 or p == 1:
        return 0
    return -p*np.log2(p) - (1-p)*np.log2(1-p)

ps = np.linspace(0.001, 0.999, 300)
H = [entropia_binaria(p) for p in ps]

plt.figure(figsize=(7,4))
plt.plot(ps, H)
plt.xlabel("p")
plt.ylabel("Entropia (bits)")
plt.title("Entropia de uma variável Bernoulli")
plt.grid(True, alpha=0.3)
plt.show()
````

# Entropia Cruzada

Se $(y)$ representa a distribuição verdadeira e $(\hat y)$ a distribuição estimada:

$$
H(y,\hat y)
=
-\sum_i y_i\log(\hat y_i)
$$

Para classificação binária:

$$
L=
-[y\log(p)+(1-y)\log(1-p)]
$$

Essa função penaliza fortemente previsões confiantes e incorretas.

Ela é uma das funções de perda mais importantes em classificação e redes neurais.

````python
def binary_cross_entropy(y, p):
    eps = 1e-12
    p = np.clip(p, eps, 1-eps)
    return -(y*np.log(p)+(1-y)*np.log(1-p))

for p in [0.1, 0.5, 0.8, 0.99]:
    print(f"y=1, p={p:.2f} -> perda={binary_cross_entropy(1,p):.4f}")
````

# Divergência de Kullback-Leibler

A divergência KL mede a diferença entre duas distribuições:

$$
D_{KL}(P\|Q)
=
\sum_x
P(x)
\log
\frac{P(x)}{Q(x)}
$$

Algumas propriedades:

- $(D_{KL}\geq0)$;
- vale zero quando as distribuições são iguais;
- não é simétrica:

$$
D_{KL}(P\|Q)\neq D_{KL}(Q\|P)
$$

Ela aparece em modelos probabilísticos, inferência variacional e aprendizado profundo.

````python
P = np.array([0.7, 0.2, 0.1])
Q = np.array([0.5, 0.3, 0.2])

kl = np.sum(P*np.log(P/Q))

print("D_KL(P || Q) =", kl)
````

# Softmax

A função Softmax transforma um vetor de valores reais em uma distribuição:

$$
P(y=i)
=
\frac{e^{z_i}}
{\sum_j e^{z_j}}
$$

Assim:

$$
\sum_iP(y=i)=1
$$

A função softmax é muito utilizada na camada de saída de classificadores multiclasse.

````python
def softmax(z):
    # subtração do máximo melhora a estabilidade numérica
    exp_z = np.exp(z-np.max(z))
    return exp_z/np.sum(exp_z)

z = np.array([2.0, 1.0, 0.1])

p = softmax(z)

print("Scores:", z)
print("Probabilidades:", p)
print("Soma:", np.sum(p))
````

# PCA (Principal Component Analysis)

A **Análise de Componentes Principais (PCA)** procura novas direções que concentrem grande parte da variabilidade dos dados.

Em linhas gerais:

```text
Dados
  │
  ▼
Centralização
  │
  ▼
Matriz de covariância
  │
  ▼
Autovalores e autovetores
  │
  ▼
Componentes principais
  │
  ▼
Projeção em menor dimensão
```

Os autovetores da matriz de covariância definem as direções principais; os autovalores indicam quanta variância está associada a cada direção.

````python
rng = np.random.default_rng(1)

x1 = rng.normal(0, 2, 200)
x2 = 0.8*x1 + rng.normal(0, 0.7, 200)

X = np.column_stack([x1,x2])

Xc = X - X.mean(axis=0)
C = np.cov(Xc, rowvar=False)

valores, vetores = np.linalg.eigh(C)
ordem = np.argsort(valores)[::-1]

valores = valores[ordem]
vetores = vetores[:,ordem]

pc1 = vetores[:,0]
projecao = Xc @ pc1

print("Autovalores:", valores)
print("Primeira componente:", pc1)

plt.figure(figsize=(6,6))
plt.scatter(X[:,0], X[:,1], alpha=0.5)
centro = X.mean(axis=0)
plt.quiver(*centro, *(pc1*4), angles="xy", scale_units="xy", scale=1)
plt.xlabel("x1")
plt.ylabel("x2")
plt.axis("equal")
plt.grid(True, alpha=0.3)
plt.show()
````

# A Matemática de um Neurônio Artificial

Um neurônio pode ser escrito como:

$$
z=\mathbf{w}^\top\mathbf{x}+b
$$

seguido por uma função de ativação:

$$
a=g(z)
$$

Portanto, um neurônio combina:

1. **Álgebra Linear** — produto escalar;
2. **Cálculo** — derivadas da função de ativação e da perda;
3. **Otimização** — ajuste de $(\mathbf w)$ e $(b)$;
4. **Probabilidade** — interpretação de algumas saídas.

````python
x = np.array([0.8, 0.2, 0.5])
w = np.array([1.1, -0.7, 0.3])
b = -0.2

z = w @ x + b
a = 1/(1+np.exp(-z))

print("z =", z)
print("sigmoid(z) =", a)
````

# A Matemática do Aprendizado por Reforço

No Aprendizado por Reforço, um agente interage com um ambiente:

```text
estado s
   │
   ▼
 agente
   │ ação a
   ▼
ambiente
   │
   ├── recompensa r
   └── próximo estado s'
```

A matemática combina probabilidade, esperança, processos de decisão e otimização.

Um conceito fundamental é o **retorno**:

$$
G_t=
R_{t+1}
+\gamma R_{t+2}
+\gamma^2R_{t+3}
+\cdots
$$

ou:

$$
G_t=
\sum_{k=0}^{\infty}
\gamma^kR_{t+k+1}
$$

onde $(0\leq\gamma\leq1)$ é o fator de desconto.

````python
recompensas = [1, 1, 1, 1, 1]
gamma = 0.9

retorno = sum((gamma**k)*r for k,r in enumerate(recompensas))

print("Recompensas:", recompensas)
print("γ =", gamma)
print("Retorno descontado =", retorno)
````

## Valor esperado

A função valor de um estado é conceitualmente:

$$
V^\pi(s)
=
E_\pi[G_t\mid S_t=s]
$$

Ou seja:

> o retorno esperado quando o agente está no estado $(s)$ e segue a política $(\pi)$.

Por isso, compreender **esperança matemática**, **probabilidade condicional** e **somatórios** é essencial para estudar RL.
