# Gradiente Descendente

Durante o treinamento, o objetivo é minimizar uma função de custo $(J(\theta))$.

Para um parâmetro:

$$
w_{novo}=w_{atual}-\alpha\frac{dJ}{dw}
$$

Em várias dimensões:

$$
\boldsymbol{\theta}_{novo}
=
\boldsymbol{\theta}_{atual}
-
\alpha\nabla J
$$

Portanto:

$$
\boxed{\nabla J\Rightarrow\text{direção/inclinação}}
\qquad
\boxed{\alpha\Rightarrow\text{tamanho do passo}}
$$

## Exemplo

Considerando:

$$
J(w)=(w-3)^2
$$

O mínimo é:

$$
w^*=3
$$

e:

$$
\frac{dJ}{dw}=2(w-3)
$$

```python
def custo(w):
    return (w-3)**2

def gradiente(w):
    return 2*(w-3)

w=np.linspace(-2,10,400)
plt.plot(w,custo(w))
plt.scatter([3],[0],s=80,label="mínimo")
plt.xlabel("w"); plt.ylabel("J(w)")
plt.title(r"$J(w)=(w-3)^2$")
plt.grid(alpha=.3); plt.legend(); plt.show()
```

## Implementação

A cada iteração:

$$
w\leftarrow w-\alpha\,2(w-3)
$$

```python
def gd(w0,alpha,iteracoes=20):
    w=float(w0)
    hist=[w]
    for _ in range(iteracoes):
        w -= alpha*gradiente(w)
        hist.append(w)
    return np.array(hist)

traj=gd(10,.1,20)
print(traj)
```

## Taxa Adequada

Começando em $(w_0=10)$ e usando:

$$
\alpha=0{,}1
$$

o parâmetro se aproxima progressivamente de $(3)$.

```python
traj=gd(10,.1,20)
w=np.linspace(-2,11,400)
plt.plot(w,custo(w))
plt.plot(traj,custo(traj),"o-",label="trajetória")
plt.scatter([3],[0],s=80,label="mínimo")
plt.xlabel("w"); plt.ylabel("J(w)")
plt.title(r"Gradiente descendente: $\alpha=0.1$")
plt.grid(alpha=.3); plt.legend(); plt.show()
```

## Taxa Muito Pequena

Com:

$$
\alpha=0{,}01
$$

os passos são pequenos.

$$
\boxed{\alpha\text{ muito pequeno}\Rightarrow\text{convergência lenta}}
$$

```python
traj=gd(10,.01,50)
plt.plot(traj,"o-",markersize=3)
plt.axhline(3,linestyle="--",label="mínimo")
plt.xlabel("Iteração"); plt.ylabel("w")
plt.title(r"Taxa pequena: $\alpha=0.01$")
plt.grid(alpha=.3); plt.legend(); plt.show()
print("Valor após 50 iterações:",traj[-1])
```

## Taxa Grande e Oscilação

Com:

$$
\alpha=0{,}9
$$

o algoritmo atravessa repetidamente o mínimo. Entretanto, a amplitude diminui.

Assim:

$$
\boxed{\text{oscilação não implica necessariamente divergência}}
$$

```python
traj=gd(10,.9,20)
plt.plot(traj,"o-")
plt.axhline(3,linestyle="--")
plt.xlabel("Iteração"); plt.ylabel("w")
plt.title(r"Oscilação: $\alpha=0.9$")
plt.grid(alpha=.3); plt.show()
print(traj[:10])
```

# Convergência

Para:

$$
J(w)=(w-3)^2
$$

temos:

$$
w_{k+1}-3=(1-2\alpha)(w_k-3)
$$

Para convergir:

$$
|1-2\alpha|<1
$$

portanto:

$$
\boxed{0<\alpha<1}
$$

**para esta função específica**.

## Caso limite: $(\alpha=1)$

Nesse caso:

$$
1-2\alpha=-1
$$

O algoritmo oscila indefinidamente entre dois pontos.

```python
traj=gd(10,1,10)
print(traj)
plt.plot(traj,"o-")
plt.axhline(3,linestyle="--")
plt.xlabel("Iteração"); plt.ylabel("w")
plt.title(r"$\alpha=1$: oscilação sem convergência")
plt.grid(alpha=.3); plt.show()
```

# Divergência

Para:

$$
\alpha>1
$$

neste exemplo:

$$
|1-2\alpha|>1
$$

e o erro aumenta.

$$
\boxed{\alpha\text{ excessivo}\Rightarrow\text{divergência}}
$$

```python
traj=gd(10,1.1,12)
plt.plot(np.abs(traj-3),"o-")
plt.yscale("log")
plt.xlabel("Iteração"); plt.ylabel(r"$|w-3|$")
plt.title(r"Divergência: $\alpha=1.1$")
plt.grid(alpha=.3); plt.show()
print(traj)
```

# Influência da Curvatura

Considere:

$$
J(w)=a(w-3)^2
$$

Então:

$$
\frac{dJ}{dw}=2a(w-3)
$$

e:

$$
w_{k+1}-3=(1-2a\alpha)(w_k-3)
$$

A condição de convergência torna-se:

$$
\boxed{0<\alpha<\frac1a}
$$

Quanto maior a curvatura $(a)$, menor pode precisar ser a taxa.

```python
def gd_curvatura(w0,alpha,a,n=20):
    w=float(w0); h=[w]
    for _ in range(n):
        w -= alpha*2*a*(w-3)
        h.append(w)
    return np.array(h)

for a in [1,5,20]:
    print("a =",a,"->",gd_curvatura(10,.1,a,5))
```

# Não é Possível Definir uma Taxa de Aprendizado Universal

O exemplo anterior mostra:

$$
\boxed{\text{não existe uma taxa de aprendizado ótima para todos os problemas}}
$$

A escolha depende da geometria da função de custo, escalas das variáveis e algoritmo utilizado.

# Duas dimensões e Zigue-Zague

Considerando a função:

$$
J(x,y)=x^2+10y^2
$$

O gradiente é:

$$
\nabla J=(2x,20y)
$$

A curvatura é muito maior na direção $(y)$, produzindo uma superfície alongada.

```python
def gd2(p0,alpha,n=30):
    p=np.array(p0,dtype=float); h=[p.copy()]
    for _ in range(n):
        g=np.array([2*p[0],20*p[1]])
        p-=alpha*g
        h.append(p.copy())
    return np.array(h)

t=gd2([4,2],.08,30)
x=np.linspace(-5,5,300); y=np.linspace(-3,3,300)
X,Y=np.meshgrid(x,y); Z=X**2+10*Y**2
plt.contour(X,Y,Z,levels=20)
plt.plot(t[:,0],t[:,1],"o-",markersize=3)
plt.scatter([0],[0],s=80,label="mínimo")
plt.xlabel("x"); plt.ylabel("y")
plt.title("Zigue-zague em uma função alongada")
plt.grid(alpha=.2); plt.legend(); plt.show()
```

# Relação de $\alpha$ com a Padronização de Dados

Se características possuem escalas muito diferentes, a função de custo pode ficar bastante alongada.

Exemplo:

- idade: $(20) – (80)$;
- renda: $(1.000) – (100.000)$.

padronizando os dados com:

$$
\boxed{z=\frac{x-\mu}{\sigma}}
$$

coloca as variáveis em escalas mais comparáveis e frequentemente melhora o comportamento numérico da otimização.

# Taxa Constante

Considerando:

$$
\alpha_k=\alpha
$$

em todas as iterações.

Há um compromisso:

- taxa maior: progresso potencialmente rápido, mas menor margem de estabilidade;
- taxa menor: maior estabilidade, mas progresso potencialmente lento.

# Decaimento da Taxa de Aprendizado

É possível reduzir a taxa durante o treinamento:

$$
\boxed{
\alpha_k=
\frac{\alpha_0}{1+\lambda k}
}
$$

A ideia é usar passos maiores no início e ajustes menores próximo ao final.

```python
alpha0=.2
lam=.05
k=np.arange(100)
a=alpha0/(1+lam*k)
plt.plot(k,a)
plt.xlabel("Iteração"); plt.ylabel("Taxa de aprendizado")
plt.title("Learning-rate decay")
plt.grid(alpha=.3); plt.show()
```

# Gradiente Descendente Estocástico

No **Batch Gradient Descent**, o gradiente é calculado com todo o conjunto de treinamento.

No **Stochastic Gradient Descent (SGD)** estrito, uma amostra é utilizada por atualização. Em redes neurais, é muito comum usar **mini-batches**.

A atualização continua sendo:

$$
\theta_{novo}
=
\theta_{atual}
-
\alpha\widehat{\nabla J}
$$

mas $(\widehat{\nabla J})$ é uma estimativa do gradiente. Por isso, a trajetória pode apresentar ruído.

# Exemplo: regressão linear

Considerando:

$$
y=3x+2+\varepsilon
$$

ajustar:

$$
\hat y=wx+b
$$

com gradiente descendente.

```python
rng=np.random.default_rng(42)
x=np.linspace(0,2,100)
y=3*x+2+rng.normal(0,.35,len(x))
plt.scatter(x,y,s=20)
plt.xlabel("x"); plt.ylabel("y")
plt.title("Dados de regressão")
plt.grid(alpha=.3); plt.show()
```

Considerando o erro quadrático médio:

$$
J(w,b)=
\frac1n\sum_{i=1}^n(\hat y_i-y_i)^2
$$

Os gradientes são:

$$
\boxed{
\frac{\partial J}{\partial w}
=
\frac2n\sum(\hat y_i-y_i)x_i
}
$$

$$
\boxed{
\frac{\partial J}{\partial b}
=
\frac2n\sum(\hat y_i-y_i)
}
$$

```python
def treinar(x,y,alpha=.01,epocas=300):
    w=b=0.0
    custos=[]
    n=len(x)
    for _ in range(epocas):
        pred=w*x+b
        erro=pred-y
        custos.append(np.mean(erro**2))
        dw=(2/n)*np.sum(erro*x)
        db=(2/n)*np.sum(erro)
        w-=alpha*dw
        b-=alpha*db
    return w,b,np.array(custos)
```

# Comparação entre Diferentes Taxas de Aprendizado na Regressão

```python
for a in [.001,.005,.01,.05]:
    _,_,custos=treinar(x,y,a,300)
    plt.plot(custos,label=f"α={a}")
plt.yscale("log")
plt.xlabel("Época"); plt.ylabel("MSE")
plt.title("Influência da taxa de aprendizado")
plt.grid(alpha=.3); plt.legend(); plt.show()
```

# Modelo Ajustado

```python
w,b,custos=treinar(x,y,.05,500)
plt.scatter(x,y,s=20,label="dados")
plt.plot(x,w*x+b,label="modelo", color='red')
plt.xlabel("x"); plt.ylabel("y")
plt.title(f"Modelo: y={w:.3f}x+{b:.3f}")
plt.grid(alpha=.3); plt.legend(); plt.show()
print("w =",w,"b =",b,"MSE =",custos[-1])
```

# Parâmetro versus Hiperparâmetro

Na regressão:

$$
w,\;b
$$

são **parâmetros aprendidos**.

Já:

$$
\alpha
$$

é um **hiperparâmetro** que controla o processo de treinamento.

# Como Escolher o Valor de $(\alpha)$?

Uma estratégia simples é testar diferentes ordens de grandeza:

$$
10^{-1},\quad10^{-2},\quad10^{-3},\quad10^{-4}
$$

e observar:

- velocidade de redução do custo;
- estabilidade;
- comportamento na validação;
- presença de oscilação ou divergência.

Também é possível utilizar *learning-rate schedules* e métodos adaptativos.

```python
taxas=[1e-4,1e-3,1e-2,1e-1]
print(f"{'alpha':>10} {'MSE inicial':>15} {'MSE final':>15}")
for a in taxas:
    _,_,custos=treinar(x,y,a,300)
    print(f"{a:10.4g} {custos[0]:15.6f} {custos[-1]:15.6f}")
```

# Otimizadores Adaptativos

Algoritmos como:

- AdaGrad;
- RMSProp;
- Adam;

ajustam efetivamente o tamanho das atualizações ao longo do treinamento.

Mesmo assim, normalmente existe uma **taxa de aprendizado base** que continua sendo importante.

# Resumo

$$
\boxed{
\theta_{novo}
=
\theta_{atual}
-
\underbrace{\alpha}_{\text{tamanho do passo }}
\underbrace{\nabla J}_{\text{direção/inclinação}}
}
$$

| Situação | Efeito típico |
|---|---|
| $(\alpha)$ muito pequeno | convergência lenta |
| $(\alpha)$ apropriado | convergência rápida e estável |
| $(\alpha)$ grande | oscilações |
| $(\alpha)$ excessivo | divergência |

A fronteira entre esses comportamentos depende do problema.
