# Análise Exploratória de Dados

O objetivo da Análise Exploratória de Dados (EDA - *Exploratory Data Analisys*) é investigar um conjunto de dados antes de aplicar modelos estatísticos ou algoritmos de aprendizado de máquina.

Em geral, a EDA ajuda a responder perguntas como:

- Quais variáveis existem no conjunto de dados?
- Qual é o tipo de cada variável?
- Existem valores ausentes?
- Existem valores extremos?
- Como os dados estão distribuídos?
- Há relações entre as variáveis?
- Os dados parecem adequados para uma análise ou modelo?

## Exemplo com o Dataset IRIS

O dataset IRIS contém medidas de flores de três espécies:

- `setosa`

As variáveis são:

- comprimento da sépala;

## Dados Ausentes

Valores ausentes são comuns em conjuntos de dados reais.  Eles podem aparecer por diferentes motivos:

- erro de coleta;
- falha de sensores;
- campos não preenchidos;
- integração incompleta de bases;
- problemas de digitação.

Antes de aplicar modelos ou análises mais profundas, é imporante verificar a presença de valores ausentes.

```python
df.isnull().sum()
```

### Injetando Dados Ausentes no Dataset IRIS

O dataset Iris não possui valores faltantes. Será criada uma cópia com dados ausentes.

```python
df_com_nulos = df.copy()

df_com_nulos.loc[0, "comprimento_sepala"] = np.nan
df_com_nulos.loc[5, "largura_petala"] = np.nan
df_com_nulos.loc[10, "especie"] = np.nan

df_com_nulos.head(12)
```

```python
df_com_nulos.isnull().sum()
```

### Visualizando Dados Ausentes com Mapa de Calor

Uma forma simples de visualizar dados faltantes é usar um mapa de calor.

```python
plt.figure(figsize=(8, 4))
sns.heatmap(df_com_nulos.isnull(), cbar=True)
plt.title("Mapa de valores ausentes")
plt.show()
```

### Estratégias para Tratar Dados Ausentes

Algumas estratégias comuns são:

| Estratégia | Quando pode ser usada |
|---|---|
| Remover linhas | Quando há poucos valores ausentes. |
| Preencher com média | Variáveis numéricas aproximadamente simétricas. |
| Preencher com mediana | Variáveis numéricas com outliers. |
| Preencher com moda | Variáveis categóricas. |
| Criar categoria "desconhecido" | Quando a ausência também carrega informação. |

A escolha da estratégia depende do problema e deve ser feita com cuidado.

```python
df_preenchido = df_com_nulos.copy()

df_preenchido["comprimento_sepala"] = df_preenchido["comprimento_sepala"].fillna(
    df_preenchido["comprimento_sepala"].mean()
)

df_preenchido["largura_petala"] = df_preenchido["largura_petala"].fillna(
    df_preenchido["largura_petala"].median()
)

df_preenchido["especie"] = df_preenchido["especie"].cat.add_categories("desconhecida")
df_preenchido["especie"] = df_preenchido["especie"].fillna("desconhecida")

df_preenchido.isnull().sum()
```

## Correlação

A correlação mede o grau de associação linear entre duas variáveis numéricas.

A correlação de Pearson varia entre -1 e 1:

| Valor | Interpretação |
|---|---|
| Próximo de 1 | Associação linear positiva forte. |
| Próximo de -1 | Associação linear negativa forte. |
| Próximo de 0 | Pouca associação linear. |

Importante: **correlação não implica causalidade**.

```python
colunas_numericas = [
    "comprimento_sepala",
    "largura_sepala",
    "comprimento_petala",
    "largura_petala"
]

correlacao = df[colunas_numericas].corr()
correlacao
```

```python
plt.figure(figsize=(8, 6))
sns.heatmap(correlacao, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Matriz de correlação")
plt.show()
```

## Identificação de Outliers com IQR

Uma técnica comum para detectar possíveis outliers usa o intervalo interquartil, ou IQR.

O IQR é calculado como:

$$
IQR = Q3 - Q1
$$

Um valor pode ser considerado possível outlier se estiver abaixo de:

$$
Q1 - 1.5 \times IQR
$$

ou acima de:

$$
Q3 + 1.5 \times IQR
$$

Essa é uma regra prática, não uma verdade absoluta.

```python
def detectar_outliers_iqr(df, coluna):
    q1 = df[coluna].quantile(0.25)
    q3 = df[coluna].quantile(0.75)
    iqr = q3 - q1

    limite_inferior = q1 - 1.5 * iqr
    limite_superior = q3 + 1.5 * iqr

    outliers = df[(df[coluna] < limite_inferior) | (df[coluna] > limite_superior)]

    return limite_inferior, limite_superior, outliers

coluna = "largura_sepala"

li, ls, outliers = detectar_outliers_iqr(df, coluna)

print("Coluna:", coluna)
print("Limite inferior:", li)
print("Limite superior:", ls)
print("Quantidade de possíveis outliers:", len(outliers))

display(outliers)
```

## Relatório de Exploração de Dados

Uma boa prática é construir um pequeno relatório exploratório com os principais pontos encontrados.

Exemplo de roteiro:

1. Dimensões do dataset.
2. Tipos de variáveis.
3. Estatísticas descritivas.
4. Valores ausentes.
5. Distribuição das variáveis.
6. Análise por grupos.
7. Correlações.
8. Possíveis outliers.
9. Hipóteses iniciais.
10. Próximos passos.

```python
def relatorio_eda(df):
    print("RELATÓRIO BÁSICO DE EDA")
    print("=" * 40)

    print("\n1. Dimensões:")
    print(f"{df.shape[0]} linhas e {df.shape[1]} colunas")

    print("\n2. Tipos de dados:")
    print(df.dtypes)

    print("\n3. Valores ausentes:")
    print(df.isnull().sum())

    print("\n4. Estatísticas descritivas:")
    display(df.describe(include="all"))

    print("\n5. Quantidade de valores únicos:")
    print(df.nunique())

relatorio_eda(df)
```

## Erros Comuns em Análise Exploratória de Dados

- Concluir causalidade apenas observando correlação.
- Ignorar valores ausentes.
- Remover outliers sem entender o motivo.
- Misturar variáveis categóricas e numéricas sem tratamento adequado.
- Não verificar se os dados estão desbalanceados.
- Usar apenas médias e ignorar dispersão.
- Fazer gráficos sem título, eixos ou contexto.
- Aplicar modelos antes de entender os dados.
