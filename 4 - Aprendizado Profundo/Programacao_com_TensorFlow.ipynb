# Definição de TensorFlow?

O **TensorFlow** é uma biblioteca de código aberto desenvolvida originalmente pelo Google para computação numérica e aprendizado de máquina.

TensorFlow é utilizada para:

- Redes neurais artificiais;
- Deep Learning;
- Visão computacional;
- Processamento de linguagem natural;
- Aprendizado por reforço profundo;
- Aplicações embarcadas e móveis com TensorFlow Lite.

A unidade fundamental de dados no TensorFlow é o **tensor**.

```python
import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt
```

# Definição de Tensores

Um **tensor** é uma estrutura de dados multidimensional.

Um tensor pode ser:

| Estrutura | Exemplo | Ordem |
|---|---|---|
| Escalar | 5 | 0D |
| Vetor | [1, 2, 3] | 1D |
| Matriz | [[1, 2], [3, 4]] | 2D |
| Tensor 3D | Imagem colorida | 3D |
| Tensor 4D | Lote de imagens | 4D |

```python
# Escalar
escalar = tf.constant(5)

# Vetor
vetor = tf.constant([1, 2, 3])

# Matriz
matriz = tf.constant([[1, 2], [3, 4]])

print("Escalar:", escalar)
print("Vetor:", vetor)
print("Matriz:\n", matriz)
```

```python
print("Shape do escalar:", escalar.shape)
print("Shape do vetor:", vetor.shape)
print("Shape da matriz:", matriz.shape)
print("Tipo da matriz:", matriz.dtype)
```

# Tensores especiais

Um TensorFlow pode ser criado e preenchido com zeros, uns, valores aleatórios e sequências numéricas.

```python
zeros = tf.zeros((3, 4))
uns = tf.ones((2, 3))
aleatorio = tf.random.normal((3, 3), mean=0, stddev=1)
sequencia = tf.range(0, 10, 2)

print("Tensor de zeros:\n", zeros)
print("\nTensor de uns:\n", uns)
print("\nTensor aleatório:\n", aleatorio)
print("\nSequência:", sequencia)
```

# Operações com Tensores

As operações matemáticas em TensorFlow são muito parecidas com operações em NumPy.

```python
a = tf.constant([1, 2, 3], dtype=tf.float32)
b = tf.constant([4, 5, 6], dtype=tf.float32)

print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("Produto escalar =", tf.reduce_sum(a * b))
```

### Multiplicação de matrizes

A multiplicação matricial é uma operação central em redes neurais.

$$
C = A \cdot B
$$

```python
A = tf.constant([[1, 2], [3, 4]], dtype=tf.float32)
B = tf.constant([[5, 6], [7, 8]], dtype=tf.float32)

C = tf.matmul(A, B)

print("A:\n", A.numpy())
print("B:\n", B.numpy())
print("A x B:\n", C.numpy())
```

# Remodelando Tensores

A forma de um tensor é muito importante em Deep Learning.

Por exemplo, imagens geralmente são representadas como:

$$
(largura, altura, canais)
$$

Um lote de imagens geralmente é representado como:

$$

# Declaração de Variáveis em TensorFlow

Um `tf.constant` representa um valor fixo.

Um `tf.Variable` representa um valor que pode ser alterado, como os pesos de uma rede neural.

```python
w = tf.Variable(2.0)

print("Valor inicial:", w.numpy())

w.assign(5.0)
print("Valor após assign:", w.numpy())

w.assign_add(3.0)
print("Valor após assign_add:", w.numpy())
```

# Cálculdos de Gradientes

Uma das principais vantagens do TensorFlow é o cálculo automático de derivadas.

Isso é essencial para treinar redes neurais usando gradiente descendente.

Suponha:

$$
y = x^2
$$

A derivada é:

$$
\frac{dy}{dx}=2x
$$

```python
x = tf.Variable(3.0)

with tf.GradientTape() as tape:
    y = x ** 2

gradiente = tape.gradient(y, x)

print("x:", x.numpy())
print("y:", y.numpy())
print("dy/dx:", gradiente.numpy())
```

# Gradiente Descendente (versão manual)

Considere a função:

$$
f(x)=x^2
$$

O mínimo dessa função ocorre em:

$$
x=0
$$

```python
x = tf.Variable(5.0)
taxa_aprendizado = 0.1
historico = []

for epoca in range(30):
    with tf.GradientTape() as tape:
        y = x ** 2

    gradiente = tape.gradient(y, x)
    x.assign_sub(taxa_aprendizado * gradiente)
    historico.append(x.numpy())

print("Valor final de x:", x.numpy())
```

```python
plt.figure(figsize=(8, 4))
plt.plot(historico, marker='o')
plt.title("Evolução de x durante o gradiente descendente")
plt.xlabel("Época")
plt.ylabel("x")
plt.grid(True)
plt.show()
```

# Regressão Linear com TensorFlow

Treinando um modelo para aproximação da função:

$$
y = 3x + 2
$$

com adição de ruído.

```python
np.random.seed(42)

X = np.linspace(-2, 2, 100).astype(np.float32)
# calculando y com ruído
y = 3 * X + 2 + np.random.normal(0, 0.3, size=X.shape).astype(np.float32)

plt.figure(figsize=(8, 4))
plt.scatter(X, y)
plt.title("Dados sintéticos")
plt.xlabel("x")
plt.ylabel("y")
plt.grid(True)
plt.show()
```

```python
modelo = tf.keras.Sequential([
    tf.keras.layers.Dense(1, input_shape=(1,))
])

modelo.compile(
    optimizer=tf.keras.optimizers.SGD(learning_rate=0.1),
    loss='mse'
)

historico = modelo.fit(X, y, epochs=50, verbose=0)
```

```python
plt.figure(figsize=(8, 4))
plt.plot(historico.history['loss'])
plt.title("Erro durante o treinamento")
plt.xlabel("Época")
plt.ylabel("MSE")
plt.grid(True)
plt.show()
```

```python
y_pred = modelo.predict(X, verbose=0)

plt.figure(figsize=(8, 4))
plt.scatter(X, y, label='Dados reais')
plt.plot(X, y_pred, label='Modelo treinado')
plt.title("Regressão linear com TensorFlow")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)
plt.show()
```

# Construindo uma Rede Neural Artificial para Classificação

Rede neural simples para classificar imagens do dataset **MNIST**.

O MNIST contém imagens de dígitos escritos à mão, de 0 a 9.

Cada imagem possui tamanho:

$$
28 \times 28
$$

```python
from tensorflow.keras.datasets import mnist

(X_train, y_train), (X_test, y_test) = mnist.load_data()

print("X_train:", X_train.shape)
print("y_train:", y_train.shape)
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
plt.figure(figsize=(4, 4))
plt.imshow(X_train[0], cmap='gray')
plt.title(f"Classe: {y_train[0]}")
plt.axis('off')
plt.show()
```

## Normalizando os Dados

Os pixels das imagens variam de 0 a 255.

Para facilitar o treinamento, é importante normalizar os valores para o intervalo entre 0 e 1.

```python
X_train = X_train.astype('float32') / 255.0
X_test = X_test.astype('float32') / 255.0
```

# Modelo com Keras Sequential

A API `Sequential` permite criar modelos camada por camada.

Neste exemplo será utilizado:

- `Flatten`: transforma a imagem 28x28 em um vetor;
- `Dense`: camada totalmente conectada;
- `ReLU`: função de ativação;
- `Softmax`: usada na saída para classificação multiclasse.

```python
modelo_mnist = tf.keras.Sequential([
    tf.keras.layers.Flatten(input_shape=(28, 28)),
    tf.keras.layers.Dense(128, activation='relu'),
    tf.keras.layers.Dense(64, activation='relu'),
    tf.keras.layers.Dense(10, activation='softmax')
])

modelo_mnist.summary()
```

# Compilando o Modelo

No processo de compilação são definidos:

- **Otimizador**: algoritmo usado para atualizar os pesos;
- **Função de perda**: mede o erro do modelo;
- **Métrica**: usada para acompanhar o desempenho.

```python
modelo_mnist.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
```

# Treinamento

Durante o treinamento, o TensorFlow realiza automaticamente:

1. Forward propagation;
2. Cálculo da perda;
3. Backpropagation;
4. Atualização dos pesos.

```python
historico_mnist = modelo_mnist.fit(
    X_train,
    y_train,
    epochs=5,
    batch_size=32,
    validation_split=0.2
)
```

# Avaliação do Modelo

```python
loss, acc = modelo_mnist.evaluate(X_test, y_test, verbose=0)

print("Loss no conjunto de teste:", loss)
print("Acurácia no conjunto de teste:", acc)
```

# Curvas de Treinamento

As curvas de treinamento ajudam a analisar:

- Convergência;
- Overfitting;
- Underfitting;
- Qualidade da generalização.

```python
plt.figure(figsize=(8, 4))
plt.plot(historico_mnist.history['loss'], label='Treino')
plt.plot(historico_mnist.history['val_loss'], label='Validação')
plt.title("Função de perda")
plt.xlabel("Época")
plt.ylabel("Loss")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
plt.plot(historico_mnist.history['accuracy'], label='Treino')
plt.plot(historico_mnist.history['val_accuracy'], label='Validação')
plt.title("Acurácia")
plt.xlabel("Época")
plt.ylabel("Acurácia")
plt.legend()
plt.grid(True)
plt.show()
```

# Fazendo Predições

```python
probabilidades = modelo_mnist.predict(X_test[:5], verbose=0)
predicoes = np.argmax(probabilidades, axis=1)

print("Predições:", predicoes)
print("Classes reais:", y_test[:5])
```

```python
plt.figure(figsize=(10, 3))

for i in range(5):
    plt.subplot(1, 5, i + 1)
    plt.imshow(X_test[i], cmap='gray')
    plt.title(f"Pred: {predicoes[i]}\nReal: {y_test[i]}")
    plt.axis('off')

plt.show()
```

# Salvando e Carregando Modelos

Após treinar um modelo é possível salvá-lo para uso posterior.

```python
modelo_mnist.save('modelo_mnist.keras')

modelo_carregado = tf.keras.models.load_model('modelo_mnist.keras')

loss, acc = modelo_carregado.evaluate(X_test, y_test, verbose=0)
print("Acurácia do modelo carregado:", acc)
```
