```python
# Script para selecionar várias imagens no disco local. Lembrando que estas imagens serão carregadas para o google drive
from google.colab import files
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt

# Carrega o modelo
model = YOLO("yolov8n.pt")

# Upload
uploaded = files.upload()

# Processa todas as imagens enviadas
for nome in uploaded.keys():
    print(f"Analisando {nome}...")

    results = model(nome)

    imagem = results[0].plot()

    plt.figure(figsize=(10,8))
    plt.imshow(imagem[..., ::-1])   # BGR -> RGB
    plt.axis("off")
    plt.show()
```

```python
# Biblioteca para carregar janela para arrastar e soltar
!pip install ultralytics gradio
```

```python
from ultralytics import YOLO
import gradio as gr

model = YOLO("yolov8n.pt")

def detectar(imagem):
    resultado = model(imagem)
    return resultado[0].plot()

interface = gr.Interface(
    fn=detectar,
    inputs=gr.Image(type="numpy"),
    outputs=gr.Image(),
    title="Detector de Objetos com YOLO",
    description="Arraste uma imagem para detectar objetos."
)

interface.launch()
```
