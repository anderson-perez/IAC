# Introdução ao Aprendizado Supervisionado

Em aprendizado supervisionado é apresentado ao algoritmo um conjunto de exemplos na forma:


$(\mathbf{x}_i, y_i)$

onde:

- $(\mathbf{x}_i)$ representa as **características de entrada** (features) do exemplo;
- $(y_i)$ representa a **resposta correta** (label/rótulo) associada àquela entrada.

O objetivo é aprender uma função:


$f(\mathbf{x}) \rightarrow y$


que seja capaz de produzir boas respostas não apenas para os exemplos usados durante o aprendizado, mas também para **novos exemplos**.

## Exemplo

| Horas de estudo | Frequência (%) | Resultado |
|---:|---:|---|
| 2 | 60 | Reprovado |
| 4 | 70 | Reprovado |
| 6 | 85 | Aprovado |
| 8 | 95 | Aprovado |

As duas primeiras colunas são características de entrada. A última coluna contém o **rótulo** conhecido. É justamente a presença dessas respostas que define aprendizado como **supervisionado**.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

dados = pd.DataFrame({
    "horas_estudo": [2, 4, 6, 8],
    "frequencia": [60, 70, 85, 95],
    "resultado": ["Reprovado", "Reprovado", "Aprovado", "Aprovado"]
})

dados
````

## Vetor de Características (X) e Alvo (objetivo) (y)

Em aprendizado de máquina É comum representar o conjunto de entradas por $(X)$ e as respostas por $(y)$.

$
X =
\begin{bmatrix}
x_{11} & x_{12} & \cdots \\
x_{21} & x_{22} & \cdots \\
\vdots & \vdots & \ddots
\end{bmatrix}
\qquad
y =
\begin{bmatrix}
y_1\\
y_2\\
\vdots
\end{bmatrix}
$

Cada **linha** de $(X)$ representa um exemplo e cada **coluna** representa uma característica.

No exemplo dos estudantes, `horas_estudo` e `frequencia` formam $(X)$, enquanto `resultado` forma $(y)$.

````python
X = dados[["horas_estudo", "frequencia"]]
y = dados["resultado"]

print("Entradas X:")
display(X)

print("\nSaídas y:")
display(y)
````

## Classificação e regressão

No aprendizado supervisionado, existem duas abordagens principais: a **classificação** e a **regressão**. Na classificação, o objetivo é determinar a qual classe ou categoria um conjunto de características pertence. Na regressão, o objetivo é prever um valor numérico contínuo a partir das características fornecidas.


### Exemplo Classificação

- aprovado / reprovado;
- normal / falha;
- gato / cachorro;
- spam / não spam.

$
f(\mathbf{x}) \rightarrow \text{classe}
$

### Exemplo Regressão

- temperatura;
- consumo de energia;
- preço de um imóvel;
- tempo de execução de uma tarefa.

$
f(\mathbf{x}) \rightarrow \mathbb{R}
$

## Classificação

Supondo um sensor de temperatura com dois possíveis valores: 

- `0` → Normal
- `1` → Alerta

É possível **intuir** uma regra:

$
\hat{y} =
\begin{cases}
0, & x < L\\
1, & x \geq L
\end{cases}
$

onde \(L\) é um **limiar**.

O símbolo $(\hat{y})$ ("y chapéu") representa a **predição** do modelo.

````python
temperaturas = np.array([20, 22, 24, 25, 27, 29, 31, 33])
rotulos =      np.array([ 0,  0,  0,  0,  1,  1,  1,  1])

limiar = 26

predicoes = (temperaturas >= limiar)

tabela = pd.DataFrame({
    "Temperatura": temperaturas,
    "Real": rotulos,
    "Predição": predicoes
})

tabela
````

### Avaliação do Algoritmo de Classificação

Uma maneira simples de avaliar esse classificador é calcular a proporção de respostas corretas:

$
\text{acurácia} =
\frac{\text{número de acertos}}
{\text{número total de exemplos}}
$

Ainda **não houve aprendizado**. O limiar ainda é escolhido manualmente.

> **Modelo** é uma regra que transforma entrada em saída.

> **Aprendizado** é o processo de ajustar essa regra utilizando dados.

````python
acuracia = np.mean(predicoes == rotulos)

print(f"Acurácia: {acuracia:.2%}")
````

### Procurando por um Limiar Automaticamente

Em vez de escolher o limiar $(L)$ manualmente, serão testados vários valores e selecionado aquele que produz o menor número de erros nos dados de treinamento.

$$
\text{dados}
\longrightarrow
\text{ajuste dos parâmetros}
\longrightarrow
\text{modelo}
$$

O único parâmetro do modelo é $(L)$.

````python
candidatos = np.arange(20, 34, 0.5)

resultados = []

for L in candidatos:
    pred = (temperaturas >= L)
    erro = np.mean(pred != rotulos)
    resultados.append((L, erro))

resultados = pd.DataFrame(resultados, columns=["Limiar", "Erro"])
melhor = resultados.loc[resultados["Erro"].idxmin()]

print(f"Melhor limiar encontrado: {melhor['Limiar']}")
print(f"Erro: {melhor['Erro']:.2%}")

plt.figure(figsize=(8, 4))
plt.plot(resultados["Limiar"], resultados["Erro"], marker="o")
plt.xlabel("Limiar")
plt.ylabel("Taxa de erro")
plt.title("Erro em função do parâmetro do modelo (limiar (L))")
plt.grid(True, alpha=0.3)
plt.show()
````

## Regressão

Considerando que $(x)$ representa o número de horas de funcionamento de uma máquina e $(y)$ o consumo acumulado de energia.

Modelo simplificado de regressão:

$$
\hat{y}=a x+b
$$

Os valores $(a)$ e $(b)$ são os **parâmetros do modelo**.

O processo de aprendizado para este modelo consiste em encontrar valores para os parâmetros $(a)$ e $(b)$ que façam as predições se aproximarem dos valores observados.

````python
x = np.array([1, 2, 3, 4, 5, 6], dtype=float)
y = np.array([3.0, 5.2, 6.8, 9.1, 10.9, 13.2])

plt.figure(figsize=(7, 4))
plt.scatter(x, y)
plt.xlabel("Horas de funcionamento")
plt.ylabel("Consumo")
plt.title("Exemplos observados")
plt.grid(True, alpha=0.3)
plt.show()
````

### Avaliação do Algoritmo de Regressão

Para cada exemplo deve ser calculado um **resíduo**:

$
e_i = y_i-\hat{y}_i
$

Uma função muito utilizada para resumir esses erros é o **Erro Quadrático Médio (MSE)**:

$
MSE = \frac{1}{n}\sum_{i=1}^{n}(y_i-\hat{y}_i)^2
$

Quanto menor o MSE, mais próximas estão as predições dos valores observados.

````python
def prever(x, a, b):
    return a*x + b

def mse(y, y_pred):
    return np.mean((y - y_pred)**2)

a = 1.5
b = 1.0

y_pred = prever(x, a, b)

print("Predições:", y_pred)
print(f"MSE: {mse(y, y_pred):.4f}")

plt.figure(figsize=(7, 4))
plt.scatter(x, y, label="Valores observados")
plt.plot(x, y_pred, label="Modelo")
plt.xlabel("Horas de funcionamento")
plt.ylabel("Consumo")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

### Localizando os Melhores Valores para os Parâmetros $(a)$ e $(b)$

Os melhores valores para os parâmetros $(a)$ e $(b)$ são aqueles que minimizam o erro.

Para encontrar esses valores, é possível realizar uma **busca exaustiva** em vários valores possíveis.

````python
melhor_erro = np.inf
melhores_parametros = None

for a in np.arange(0, 4.01, 0.05):
    for b in np.arange(-2, 4.01, 0.05):
        y_pred = prever(x, a, b)
        erro = mse(y, y_pred)

        if erro < melhor_erro:
            melhor_erro = erro
            melhores_parametros = (a, b)

a_melhor, b_melhor = melhores_parametros

print(f"a = {a_melhor:.2f}")
print(f"b = {b_melhor:.2f}")
print(f"MSE = {melhor_erro:.4f}")

y_pred = prever(x, a_melhor, b_melhor)

plt.figure(figsize=(7, 4))
plt.scatter(x, y, label="Dados")
plt.plot(x, y_pred, label="Modelo ajustado")
plt.xlabel("Horas de funcionamento")
plt.ylabel("Consumo")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

## Separando as Etapas: Treinamento e Teste

Tanto nos problemas de **classificação** quanto nos de **regressão**, o desempenho do modelo não deve ser avaliado utilizando apenas os mesmos dados empregados durante o seu treinamento. Para verificar se o modelo é capaz de **generalizar para novos exemplos**, é necessário dividir os dados disponíveis em, pelo menos, dois conjuntos: o **conjunto de treinamento**, utilizado para ajustar o modelo, e o **conjunto de teste**, utilizado para avaliar seu desempenho com dados que não foram utilizados durante o treinamento.

````python
rng = np.random.default_rng(42)

x = np.arange(1, 21, dtype=float)
y = 2*x + 3 + rng.normal(0, 2, size=len(x))

# Separação manual apenas para fins didáticos
x_treino, y_treino = x[:14], y[:14]
x_teste,  y_teste  = x[14:], y[14:]

print("Exemplos de treinamento:", len(x_treino))
print("Exemplos de teste:", len(x_teste))
````

````python
melhor_erro = np.inf
melhores_parametros = None

for a in np.arange(0, 4.01, 0.05):
    for b in np.arange(-1, 7.01, 0.1):
        pred_treino = prever(x_treino, a, b)
        erro = mse(y_treino, pred_treino)

        if erro < melhor_erro:
            melhor_erro = erro
            melhores_parametros = (a, b)

a, b = melhores_parametros

pred_treino = prever(x_treino, a, b)
pred_teste = prever(x_teste, a, b)

print(f"Modelo aprendido: y = {a:.2f}x + {b:.2f}")
print(f"MSE de treinamento: {mse(y_treino, pred_treino):.3f}")
print(f"MSE de teste:       {mse(y_teste, pred_teste):.3f}")

plt.figure(figsize=(8, 4))
plt.scatter(x_treino, y_treino, label="Treinamento")
plt.scatter(x_teste, y_teste, label="Teste")
plt.plot(x, prever(x, a, b), label="Modelo")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

## Generalização

O objetivo de um sistema de aprendizado supervisionado **não é simplesmente memorizar o conjunto de treinamento**. Ele necessita aprender padrões que possam ser utilizados em novas situações:

```text
Exemplos conhecidos
       │
       ▼
   Treinamento
       │
       ▼
     Modelo
       │
       ▼
Exemplo nunca visto
       │
       ▼
    Predição
```

Se o desempenho é bom apenas nos dados de treinamento, mas ruim em novos dados, o modelo não está generalizando adequadamente.

## Underfitting, Overfitting e Modelo Adequado

O underfitting e o overfitting são características de modelos com problemas de generalização.

### Underfitting

O modelo não consegue representar adequadamente o padrão existente nos dados.

### Overfitting

O modelo passa a representar até mesmo ruídos e particularidades do conjunto de treinamento, podendo apresentar desempenho ruim em novos exemplos.

### Modelo Adequado

É capaz de aprender os padrões presentes nos dados de treinamento e generalizar esse conhecimento para novos exemplos.

````python
rng = np.random.default_rng(7)

x = np.linspace(-3, 3, 15)
y_real = 0.5*x**2 - x + 2
y = y_real + rng.normal(0, 1.2, size=len(x))

x_curva = np.linspace(-3.2, 3.2, 300)

# Apenas para ilustração matemática:
coef_simples = np.polyfit(x, y, 1)
coef_adequado = np.polyfit(x, y, 2)
coef_complexo = np.polyfit(x, y, 12)

for titulo, coef in [
    ("Underfitting: grau 1", coef_simples),
    ("Ajuste adequado: grau 2", coef_adequado),
    ("Possível overfitting: grau 12", coef_complexo)
]:
    plt.figure(figsize=(7, 4))
    plt.scatter(x, y, label="Dados")
    plt.plot(x_curva, np.polyval(coef, x_curva), label=titulo)
    plt.xlabel("x")
    plt.ylabel("y")
    plt.title(titulo)
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
````

## Passos Necessário para se Obter um bom Modelo em Aprendizado Supervisionado

O aprendizado supervisionado pode ser resumido da seguinte forma:

```text
Dados rotulados
      │
      ▼
Separação das entradas X e respostas y
      │
      ▼
Divisão em treinamento e teste
      │
      ▼
Definição de um modelo f(X; θ)
      │
      ▼
Predições ŷ
      │
      ▼
Comparação entre y e ŷ
      │
      ▼
Cálculo do erro
      │
      ▼
Ajuste dos parâmetros θ
      │
      ▼
Modelo treinado
      │
      ▼
Avaliação com dados não vistos
      │
      ▼
Generalização
```

O $(\theta)$ é utilizado para representar coletivamente os **parâmetros que o modelo precisa aprender**.
